```python
import numpy as np
import pandas as pd
```

```python
matches=pd.read_csv('matches.csv')
delivery=pd.read_csv('deliveries.csv')
```

```python
matches.head()
```

```python
delivery.head()
```

```python
total_score_df=delivery.groupby(['match_id','inning']).sum()['total_runs'].reset_index()
```

```python
total_score_df=total_score_df[total_score_df['inning']==1]
```

```python
total_score_df.head()
```

```python
match_df=matches.merge(total_score_df[['match_id','total_runs']],left_on='id',right_on='match_id')
```

```python
match_df['team1'].unique()
```

```python
teams=[
    'Royal Challengers Bengaluru',
    'Mumbai Indians',
    'Kolkata Knight Riders',
    'Rajasthan Royals', 
    'Delhi Capitals', 
    'Punjab Kings',
    'Lucknow Super Giants',
    'Gujarat Titans', 
    'Sunrisers Hyderabad',
    'Chennai Super Kings'
]
```

```python
match_df['team1']=match_df['team1'].str.replace('Royal Challengers Bangalore','Royal Challengers Bengaluru')
match_df['team2']=match_df['team2'].str.replace('Royal Challengers Bangalore','Royal Challengers Bengaluru')
```

```python
match_df['team1']=match_df['team1'].str.replace('Kings XI Punjab','Punjab Kings')
match_df['team2']=match_df['team2'].str.replace('Kings XI Punjab','Punjab Kings')
```

```python
match_df['team1']=match_df['team1'].str.replace('Deccan Chargers','Sunrisers Hyderabad')
match_df['team2']=match_df['team2'].str.replace('Deccan Chargers','Sunrisers Hyderabad')
```

```python
match_df['team1']=match_df['team1'].str.replace('Delhi Daredevils','Delhi Capitals')
match_df['team2']=match_df['team2'].str.replace('Delhi Daredevils','Delhi Capitals')
```

```python
match_df=match_df[match_df['team1'].isin(teams)]
match_df=match_df[match_df['team2'].isin(teams)]
```

```python
match_df.head()
```

```python
match_df=match_df[match_df['method']!='D/L']
match_df['method'].value_counts()
```

```python
match_df = match_df[['match_id', 'city', 'winner', 'total_runs']]
delivary_df=match_df.merge(delivery,on='match_id')
```

```python
delivary_df=delivary_df[delivary_df['inning']==2]
```

```python
delivary_df.head()
```

```python
delivary_df['current_score'] = delivary_df.groupby('match_id')['total_runs_y'].cumsum()
```

```python
delivary_df['runs_left']=1+delivary_df['total_runs_x']-delivary_df['current_score']
```

```python
delivary_df['balls_left']=(120)-((delivary_df['over'])*6+delivary_df['ball'])
```

```python
delivary_df['wicket_fall']=delivary_df.groupby('match_id')['is_wicket'].cumsum()
```

```python
delivary_df['wickets_left']=10 - delivary_df['wicket_fall']
```

```python
delivary_df.head()
```

```python
delivary_df['crr']=delivary_df['current_score']*6/(120 - delivary_df['balls_left'])
```

```python
delivary_df['rrr']=(delivary_df['runs_left'])*6/delivary_df['balls_left']
```

```python
delivary_df['result']=(delivary_df['winner']==delivary_df['batting_team']).astype(int)
```

```python
final_df=delivary_df[[
    'match_id',
    'batting_team', 'bowling_team', 'city', 'runs_left',
    'balls_left', 'wickets_left', 'total_runs_x',
    'crr', 'rrr', 'result'
]]
```

```python
final_df=final_df.sample(final_df.shape[0])
```

```python
teams = [
    'Royal Challengers Bengaluru',
    'Mumbai Indians',
    'Kolkata Knight Riders',
    'Rajasthan Royals',
    'Delhi Capitals',
    'Punjab Kings',
    'Lucknow Super Giants',
    'Gujarat Titans',
    'Sunrisers Hyderabad',
    'Chennai Super Kings'
]

final_df['batting_team'] = final_df['batting_team'].str.replace('Royal Challengers Bangalore', 'Royal Challengers Bengaluru')
final_df['bowling_team'] = final_df['bowling_team'].str.replace('Royal Challengers Bangalore', 'Royal Challengers Bengaluru')

final_df['batting_team'] = final_df['batting_team'].str.replace('Kings XI Punjab', 'Punjab Kings')
final_df['bowling_team'] = final_df['bowling_team'].str.replace('Kings XI Punjab', 'Punjab Kings')

final_df['batting_team'] = final_df['batting_team'].str.replace('Deccan Chargers', 'Sunrisers Hyderabad')
final_df['bowling_team'] = final_df['bowling_team'].str.replace('Deccan Chargers', 'Sunrisers Hyderabad')

final_df['batting_team'] = final_df['batting_team'].str.replace('Delhi Daredevils', 'Delhi Capitals')
final_df['bowling_team'] = final_df['bowling_team'].str.replace('Delhi Daredevils', 'Delhi Capitals')

final_df = final_df[final_df['batting_team'].isin(teams)]
final_df = final_df[final_df['bowling_team'].isin(teams)]
```
